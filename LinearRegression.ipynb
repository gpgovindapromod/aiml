```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error  # Add this line

# Load the preprocessed DataFrame from Google Drive
df = pd.read_csv('/content/drive/MyDrive/titanic_preprocessed.csv')

# Select Age and Fare columns and remove missing values
df = df[['Age', 'Fare']].dropna()

# Independent variable (X)
X = df[['Age']]

# Dependent variable (Y)
Y = df[['Fare']]

# Train the model
model = LinearRegression()
model.fit(X, Y)

# Predict values
Y_pred = model.predict(X)

# Plot
plt.figure(figsize=(8,6))
plt.scatter(X, Y, color='blue', label='Data Points')
plt.plot(X, Y_pred, color='red', linewidth=2, label='Regression Line')
plt.title('Simple Linear Regression: Age vs Fare (Titanic Dataset)')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend()
plt.grid(True)
plt.show()

# Print model parameters
print("Slope (Coefficient):", model.coef_[0][0])
print("Intercept:", model.intercept_[0])

# Calculate R² Score and Mean Squared Error
r2 = r2_score(Y, Y_pred)
mse = mean_squared_error(Y, Y_pred)

print("R² Score:", r2)
print("Mean Squared Error:", mse)
```

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

# Load the preprocessed DataFrame from Google Drive
df = pd.read_csv('/content/drive/MyDrive/titanic_preprocessed.csv')

# Select required columns and remove missing values
df = df[['Age', 'Pclass', 'Fare']].dropna()

# Input features (Independent variables)
X = df[['Age', 'Pclass']]

# Target variable (Dependent variable)
Y = df[['Fare']]

# Train the multiple linear regression model
model = LinearRegression()
model.fit(X, Y)

# Predict values
Y_pred = model.predict(X)

# Print model parameters
print("Coefficients:", model.coef_[0])
print("Intercept:", model.intercept_[0])

# Calculate R² Score and Mean Squared Error
r2 = r2_score(Y, Y_pred)
mse = mean_squared_error(Y, Y_pred)

print("R² Score:", r2)
print("Mean Squared Error:", mse)
```
