```python
# Uniform Cost search
import heapq

def uniform_cost_search(graph, start, goal):  #graph → The weighted graph.
                                              #start → Starting node.
                                              #goal → Destination node.
    # Priority queue: (cost, current_node, path)
    queue = [(0, start, [])]    #cost: 0;   current node: start;    path: empty
    visited = set()

    while queue:
        cost, node, path = heapq.heappop(queue) #heapq.heappop() removes the tuple with the smallest cost.

        if node in visited:
            continue

        visited.add(node)
        path = path + [node]

        if node == goal:
            return cost, path

        for neighbor, edge_cost in graph.get(node, []):
            if neighbor not in visited:
                heapq.heappush(queue, (cost + edge_cost, neighbor, path))

    return float("inf"), []

# Graph input
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 3), ('C', 2)],
    'C': [('E', 5)],
    'D': [('F', 2) , ('G', 4)],
    'E': [('G', 3)],
    'F': [('G', 1)],
    'G': []
}

start = 'A'
goal = 'G'

cost, path = uniform_cost_search(graph, start, goal)

print("Shortest Cost:", cost)
print("Path:", " -> ".join(path))
```

```python
# Astar

import heapq

# Heuristic values given by user
heuristic = {
    'A': 6,
    'B': 4,
    'C': 4,
    'D': 2,
    'E': 2,
    'F': 2,
    'G': 0
}

# Graph: Node -> (Neighbor, Cost)
graph = {
    'A': [('B', 2.2), ('C', 2.2)],
    'B': [('D', 2.2), ('E', 3.2)],
    'C': [('F', 3.2)],
    'D': [('G', 3.0)],
    'E': [('G', 2.2)],
    'F': [('G', 2.2)],
    'G': []
}


# A* Algorithm
def astar(graph, start, goal):

    open_list = []
    heapq.heappush(open_list, (0, start))

    g = {start: 0}
    parent = {start: None}
    visited = set()

    while open_list:

        current_f, current = heapq.heappop(open_list)

        if current in visited:
            continue

        visited.add(current)

        print(f"\nCurrent Node: {current}")

        # Goal reached
        if current == goal:
            path = []

            while current:
                path.append(current)
                current = parent[current]

            return path[::-1], g[goal]

        # Check neighbors
        for neighbor, cost in graph[current]:

            if neighbor in visited:
                continue

            # g(n)
            g_new = g[current] + cost

            # h(n) is given by user
            h_new = heuristic[neighbor]

            # f(n) = g(n) + h(n)
            f_new = g_new + h_new

            if neighbor not in g or g_new < g[neighbor]:

                g[neighbor] = g_new
                parent[neighbor] = current

                heapq.heappush(open_list, (f_new, neighbor))

                print(f"Node: {neighbor}")
                print(f"g(n) = {g_new:.2f}")
                print(f"h(n) = {h_new:.2f}")
                print(f"f(n) = g(n) + h(n) = {g_new:.2f} + {h_new:.2f} = {f_new:.2f}")
                print("-" * 30)

    return None, float("inf")


# Main Program
start = 'A'
goal = 'G'

path, cost = astar(graph, start, goal)

print("\nOptimal Path:", " -> ".join(path))
print("Total Cost:", round(cost, 2))
```
